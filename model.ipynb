# NOTE : Data Cleaning of whole datasets (PHASE)

```python
import pandas as pd
import numpy as np


# NOTE  Standardize text: removing whitespaces from beginning and ending 
df['model'] = df['model'].astype(str).str.lower().str.strip()
#NOTE: removing non - alphanumeric characters using regex [sajilo raixa]
df['model'] = df['model'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True).str.strip()

#NOTE : handling missing values [by replacing the emptyStrings with NaN(Actual)]
df['model'] = df['model'].replace('', np.nan)
#NOTE: again filling those NaN with category name 'others'
df['model'] = df['model'].fillna('other')

#NOTE:  Target and replace ONLY rare models safely
model_counts = df['model'].value_counts()
rare_models = model_counts[model_counts < 10].index
df.loc[df['model'].isin(rare_models), 'model'] = 'other'


print(df['model'].value_counts().head(10))
```

# NOTE: Feature Engineering (phase)

```python
# #NOTE: Target encoding applying for col ['model']
from category_encoders import TargetEncoder

TE_model = TargetEncoder(cols=['model'])
TE_manufacturer = TargetEncoder(cols=['manufacturer'])

#NOTE: we only fit_transform X_train only
X_train['model_Encoded'] = TE_model.fit_transform(X_train['model'], y_train)
X_train['manufacturer_Encoded'] = TE_manufacturer.fit_transform(X_train['manufacturer'], y_train)

#NOTE: we tranform X_test only
X_test['model_Encoded'] = TE_model.transform(X_test['model'])
X_test['manufacturer_Encoded'] = TE_manufacturer.transform(X_test['manufacturer'])

#Now , droppping the original one since now , we don't need after sucessfull encoding 
X_train.drop(columns=['model', 'manufacturer'], inplace=True)
X_test.drop(columns=['model', 'manufacturer'], inplace=True)
```

```python
X_train['model_Encoded'] #NOTE: now cars are assigned with the avg. price of their model names
```

```python
X_train['manufacturer_Encoded']
```

```python
#NOTE: Applying ordinal encoding on ['condition'] since the values are ordinal
# using sequence :condition_order = ['salvage', 'fair', 'good', 'excellent', 'like new', 'new']
# We need to assign from 0 to 1,2..... increasing order

from sklearn.preprocessing import OrdinalEncoder
OE = OrdinalEncoder()
```

```python
order = ['salvage', 'fair', 'good', 'excellent', 'like new', 'new']
OE = OrdinalEncoder(
    categories=[order]
)

#NOTE : here we have passed in 2D since among X_train's many col , we only want to apply on 1 particular col called 'condition'
X_train['condition_Encoded'] = OE.fit_transform(X_train[['condition']])
X_test['condition_Encoded'] = OE.transform(X_test[['condition']])

# Drop the original text column afterward (dropping the original one since we don't need them now)
X_train.drop(columns=['condition'], inplace=True)
X_test.drop(columns=['condition'], inplace=True)
```

```python
X_train
```

```python
#NOTE: now , handling [fuel] by applying one hot encoding
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(
    drop='first',
    sparse_output=False,
    handle_unknown='ignore'
)

X_train_fuel_Encoded = ohe.fit_transform(X_train[['fuel']])
X_test_fuel_Encoded = ohe.transform(X_test[['fuel']])

#Getting new encoded names of features 
Encoded_Names = ohe.get_feature_names_out(['fuel'])

#NOTE: now we need to convert back into dataframes and map the new feature names 
X_trainFinal = pd.DataFrame(X_train_fuel_Encoded, columns=Encoded_Names, index=X_train.index)
X_testFinal = pd.DataFrame(X_test_fuel_Encoded, columns=Encoded_Names, index=X_test.index)

#NOTE: finally,we need to concatenate newly created columns to our main X_train and drop the original one
X_train = pd.concat([X_train, X_trainFinal], axis=1).drop(columns=['fuel'])
X_test = pd.concat([X_test, X_testFinal], axis=1).drop(columns=['fuel'])

```

```python
X_train #NOTE: we can see OneHotEncoder performed well
```

```python
#NOTE: OneHotEncoding for column ['title_status']

ohe_status = OneHotEncoder(
    drop='first',
    sparse_output=False,
    handle_unknown='ignore'
)

X_train_status_Encoded = ohe_status.fit_transform(X_train[['title_status']])
X_test_status_Encoded = ohe_status.transform(X_test[['title_status']])

Encoded_StatusNames = ohe_status.get_feature_names_out(['title_status'])

X_train_status_final = pd.DataFrame(X_train_status_Encoded, columns=Encoded_StatusNames,index=X_train.index)
X_test_status_final = pd.DataFrame(X_test_status_Encoded, columns=Encoded_StatusNames,index=X_test.index)

X_train = pd.concat([X_train, X_train_status_final], axis=1).drop(columns=['title_status'])
X_test = pd.concat([X_test, X_test_status_final], axis=1).drop(columns=['title_status'])
```

```python
X_train
```

```python
#NOTE: yet, we have to apply OneHotEncoding for column 'type' and 'transmission'
# since the previous method i was using to do individually is taking me too long so,
# now, i am gonna use concept of column transformer in order to decrease the code size and efficieny 

from sklearn.compose import ColumnTransformer

CT = ColumnTransformer(
    transformers=[('OHE_type', OneHotEncoder(), ['type']),
    ('OHE_transmission', OneHotEncoder(), ['transmission'])],
    remainder='passthrough'
)

Transformed_X_train = CT.fit_transform(X_train) #by default Columntransformer return numpy array
Transformed_X_test = CT.transform(X_test)

#NOTE: Extracting the transformed names we got after applying [OneHotEncoder]
transformed_names = CT.get_feature_names_out() #NOTE: unlike previous method , here we don't have to pass cols names list

#NOTE: Converting Numpy arrays() back to pandas dataframes
X_train = pd.DataFrame(Transformed_X_train, columns=transformed_names, index=X_train.index)
X_test = pd.DataFrame(Transformed_X_test, columns=transformed_names, index=X_test.index)

#NOTE : Since case of applying column transformer, i don't have to concat the transformed features unlike the 
# explicit methods and also i don't have to drop the original column , because column transformer handles 
# itself, internally
```
