```python
#NOTE: now , checking VIF of each categorical numeric columns 
from statsmodels.stats.outliers_influence import variance_inflation_factor


# 1. 
# Ensure you have handled NaNs (imputed) before this step!
numeric_df = df.select_dtypes(include=['int64', 'float64']).dropna()

# 2. VIF needs an explicit constant/intercept column to calculate correctly
# (statsmodels doesn't add this automatically)
X = numeric_df.copy()
X['intercept'] = 1

# Calculate VIF for each feature
vif_data = pd.DataFrame()   #NOTE: creating a empty dataframes
vif_data["feature"] = numeric_df.columns    #NOTE: passing all numeric dataframes into new empty df
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(numeric_df.shape[1])] #This is the native 
#statsmodels function. In each round of the loop, it treats column i as the temporary target, runs a regression 
# against all other columns inside the X.values matrix, and spits out that column's individual VIF score.
#since VIF removes orignal target temporarily and calculates R2 using all independent features

# 4. Sort from highest VIF to lowest VIF
vif_data = vif_data.sort_values(by="VIF", ascending=False).reset_index(drop=True) #since after resuffling the order
# gonna change so , resetting it too

print("=== VIF Results ===")
print(vif_data)
```

# Imputation of missing values 

```python
from sklearn.impute import SimpleImputer

# For categorical columns (condition, type)
# Use most_frequent — fills with the most common category
category_imputer = SimpleImputer(strategy='most_frequent')
X_train[['condition', 'type']] = category_imputer.fit_transform(X_train[['condition', 'type']])
X_test[['condition', 'type']] = category_imputer.transform(X_test[['condition', 'type']])


# For numeric columns (cylinders if stored as number) -- cause i have stored it as float
# Use median — more robust than mean when data is skewed
numerical_imputer = SimpleImputer(strategy='median')
X_train[['cylinders']] = numerical_imputer.fit_transform(X_train[['cylinders']])
X_test[['cylinders']] = numerical_imputer.transform(X_test[['cylinders']])
```

```python
X_train.isnull().sum()
```

```python
X_test.isnull().sum()
```

```python
# #NOTE: Target encoding applying for col ['model']
from category_encoders import TargetEncoder

TE_model = TargetEncoder(cols=['model'])
TE_manufacturer = TargetEncoder(cols=['manufacturer'])

#NOTE: we only fit_transform X_train only
X_train['model_Encoded'] = TE_model.fit_transform(X_train['model'], y_train)
X_train['manufacturer_Encoded'] = TE_manufacturer.fit_transform(X_train['manufacturer'], y_train)

#NOTE: we tranform X_test only
X_test['model_Encoded'] = TE_model.transform(X_test['model'])
X_test['manufacturer_Encoded'] = TE_manufacturer.transform(X_test['manufacturer'])

#Now , droppping the original one since now , we don't need after sucessfull encoding 
X_train.drop(columns=['model', 'manufacturer'], inplace=True)
X_test.drop(columns=['model', 'manufacturer'], inplace=True)
```

```python
#NOTE: Applying ordinal encoding on ['condition'] since the values are ordinal
# using sequence :condition_order = ['salvage', 'fair', 'good', 'excellent', 'like new', 'new']
# We need to assign from 0 to 1,2..... increasing order

from sklearn.preprocessing import OrdinalEncoder
OE = OrdinalEncoder()
```

```python
order = ['salvage', 'fair', 'good', 'excellent', 'like new', 'new']
OE = OrdinalEncoder(
    categories=[order]
)

#NOTE : here we have passed in 2D since among X_train's many col , we only want to apply on 1 particular col called 'condition'
X_train['condition_Encoded'] = OE.fit_transform(X_train[['condition']])
X_test['condition_Encoded'] = OE.transform(X_test[['condition']])

# Drop the original text column afterward (dropping the original one since we don't need them now)
X_train.drop(columns=['condition'], inplace=True)
X_test.drop(columns=['condition'], inplace=True)
```

```python
X_train
```

```python
#NOTE: now , handling [fuel] by applying one hot encoding
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(
    drop='first',
    sparse_output=False,
    handle_unknown='ignore'
)

X_train_fuel_Encoded = ohe.fit_transform(X_train[['fuel']])
X_test_fuel_Encoded = ohe.transform(X_test[['fuel']])

#Getting new encoded names of features 
Encoded_Names = ohe.get_feature_names_out(['fuel'])

#NOTE: now we need to convert back into dataframes and map the new feature names 
X_trainFinal = pd.DataFrame(X_train_fuel_Encoded, columns=Encoded_Names, index=X_train.index)
X_testFinal = pd.DataFrame(X_test_fuel_Encoded, columns=Encoded_Names, index=X_test.index)

#NOTE: finally,we need to concatenate newly created columns to our main X_train and drop the original one
X_train = pd.concat([X_train, X_trainFinal], axis=1).drop(columns=['fuel'])
X_test = pd.concat([X_test, X_testFinal], axis=1).drop(columns=['fuel'])

```

```python
X_train #NOTE: we can see OneHotEncoder performed well
```

```python
#NOTE: yet, we have to apply OneHotEncoding for column 'type' and 'transmission'
# since the previous method i was doing individually and it was taking me too long so,
# now, i am gonna use concept of column transformer in order to decrease the code size and efficieny 

from sklearn.compose import ColumnTransformer

CT = ColumnTransformer(
    transformers=[('OHE_type', OneHotEncoder(), ['type']),
    ('OHE_transmission', OneHotEncoder(), ['transmission'])],
    remainder='passthrough'
)

Transformed_X_train = CT.fit_transform(X_train) #by default Columntransformer return numpy array
Transformed_X_test = CT.transform(X_test)

#NOTE: Extracting the transformed names we got after applying [OneHotEncoder]
transformed_names = CT.get_feature_names_out() #NOTE: unlike previous method , here we don't have to pass cols names list

#NOTE: Converting Numpy arrays() back to pandas dataframes
X_train = pd.DataFrame(Transformed_X_train, columns=transformed_names, index=X_train.index)
X_test = pd.DataFrame(Transformed_X_test, columns=transformed_names, index=X_test.index)

#NOTE : Since case of applying column transformer, i don't have to concat the transformed features unlike the 
# explicit methods and also i don't have to drop the original column , because column transformer handles 
# itself, internally
```

# NOW : FEATURE SCALING [PHASE] - Actually easy bro!

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

#NOTE: since scaler returns numpy array so, we need to convert it back to pandas DataFrame
X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
X_test = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)
```

# MODEL TRAINING :

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

models = {
    'LinearRegression' : LinearRegression(),
    'Ridge'            : Ridge(alpha=1.0),
    'Lasso'            : Lasso(alpha=0.001),
    'RandomForest'     : RandomForestRegressor(n_estimators=100, random_state=42)
}

results = {}

for name, model in models.items(): #NOTE: harek 4 ota model ko laggi fit hunxa , predict garxa and then result lai
    model.fit(X_train, y_train) #store garxa in a dictionary ma nested tarika ley that is all model ko info haru
    y_pred = model.predict(X_test)
    
    results[name] = {
        'R2'   : r2_score(y_test, y_pred),
        'MAE'  : mean_absolute_error(y_test, y_pred),
        'RMSE' : np.sqrt(mean_squared_error(y_test, y_pred))
    }

results_df = pd.DataFrame(results).T #NOTE: converting result into a dataframe strucutre 
# .T ==> transposing 
print(results_df.round(4))
```

```python
#NOTE: Since, we got best model as 'RandomForest' and also still the scale is in log but we need to convert into
# actual $ salary for better interpretability

best_model = models['RandomForest']
y_pred_log = best_model.predict(X_test)

#converting the scale
y_pred_dollar = np.expm1(y_pred_log)
y_test_dollar = np.expm1(y_test)

print(f"R2    : {r2_score(y_test_dollar, y_pred_dollar):.4f}")
print(f"MAE   : ${mean_absolute_error(y_test_dollar, y_pred_dollar):,.0f}")
print(f"RMSE  : ${np.sqrt(mean_squared_error(y_test_dollar, y_pred_dollar)):,.0f}")
```
